```python
# 1. Definimos nuestros datos de entrada (Variables) ingreso_mensual = 500000 alquiler
= 180000 servicios = 45000 comida = 120000 # 2. La computadora realiza los cálculos
matemáticos gastos_totales = alquiler + servicios + comida dinero_libre = ingreso_mensual -
gastos_totales # 3. Mostramos los resultados por pantalla print("--- RESUMEN DE TU MES ---")
print("Tus gastos totales son: $", gastos_totales) print("Te quedan libres para disfrutar:
$", dinero_libre)
```

```python
# Evaluamos la cantidad de dinero libre que nos queda if dinero_libre > 100000:
print("🎉 ¡Excelente mes! Puedes guardar un poco para ahorrar o darte un gusto.") elif
dinero_libre > 0: print("⚠️ Llegas a fin de mes, pero cuidado con los gastos extras.") else:
print("🚨 ¡Atención! Estás gastando más de lo que ingresa. Hay que ajustar el presupuesto.")
```

```python
sueldo_actual = 500000 porcentaje_aumento = 1.10 # Representa un aumento del 10%
print("--- PROYECCIÓN DE SUELDO A 6 MESES ---") # Repetimos el cálculo automáticamente
durante 6 meses for mes in range(1, 7): sueldo_actual = sueldo_actual * porcentaje_aumento
print("Mes", mes, ": Ganarás $", round(sueldo_actual))
```
